## Introduction

### Background and Context
The Titanic disaster is one of the most infamous maritime tragedies in history. On April 15, 1912, the RMS Titanic sank after colliding with an iceberg during its maiden voyage from Southampton to New York City. Out of the approximately 2,224 passengers and crew on board, up to 1,635 lost their lives[¹](#reference_01). The disaster has since become a symbol of human error and the consequences of inadequate safety measures.

Kaggle, a platform for data science competitions, has hosted a popular competition called "[Titanic: Machine Learning from Disaster](https://www.kaggle.com/competitions/titanic)." This notebook is an attempt to get a reasonable score on the Titanic Kaggle competition by applying machine learning techniques to predict the survival of passengers based on various features.

### Problem Statement
The primary problem we aim to solve is to predict whether a passenger survived the Titanic disaster based on the features provided in the dataset. The dataset includes information such as the passenger's age, gender, ticket class, and other relevant details. By building a machine learning model, we can analyze the factors that influenced survival and make informed predictions about the likelihood of survival for each passenger.


### Objectives & Success Criteria

1. **Primary Objective:** To build a machine learning model that can predict whether a passenger survived the Titanic disaster based on the provided dataset.
2. **Secondary Objective:** To provide insights into the factors that influenced survival.
3. **Business Success Criteria:** The model should correctly classify the survival outcome of passengers at a level that demonstrates practical predictive value.
4. **Technical Success Criteria:**
    * **Primary Metric:** Accuracy *(the proportion of correct predictions -both true positives and true negatives- out of the total number of predictions made)* will be used as the primary evaluation metric. Even if an imbalanced dataset is revealed during the Exploratory Data Analysis (EDA), accuracy will still be the primary metric for evaluation because this is how the Kaggle competition evaluates submissions. 
    * **Acceptance Threshold:** An accuracy of 80% or higher on the test set will be considered a successful outcome for this project.

### Solution Requirements

* **Output:** A CSV file containing the predicted survival status for each passenger in the test set.
* **Delivery:** The submission will be made through the Kaggle competition platform. The final model and code will be documented and shared in a Jupyter Notebook format for reproducibility.
* **Interpretability:** Interpretability of the model is not a primary requirement for the competition, therefore a black-box model will be preferred if it provides significantly better performance. However, if the champion model and the runner-up model give similar performance, the more interpretable model will be preferred even if it is not the champion model.
* **Cadence:** This project will be completed in a single phase, with the final submission made at the end of the project timeline. No future real-world data will be fed into the model after the submission, as the competition is based on a static dataset.

### Analytical Approach

* **Problem Type:** Supervised Binary Classification.
* **Target Variable:** `survived` (1 for survived, 0 for did not survive).
* **Justification:** 
    * The problem is a binary classification task because we are trying to predict one of two possible outcomes (survived or did not survive) based on the features provided in the dataset.
    * This is a supervised learning problem because we have labeled data (the training set) that we can use to train our model.

**Candidate Models:**

| # | Algorithm | Rationale |
| :--- | :--- | :--- |
| 1 (Baseline) | Logistic Regression | Fast, computationally cheap, and highly interpretable - the coefficients directly reveal the direction and strength of each feature's influence on the predicted class. Ideal when a reliable first threshold before trying more complex classifiers is needed. |
| 2 | Random Forest | A good candidate for general-purpose classification and regression. Generally provides high accuracy without much tuning. Excellent default choice when interpretability is not the top priority. Robust to overfitting, handles mixed feature types, and provides built-in feature importance rankings. |
| 3 | XGBoost | A good candidate for competition-winning accuracy on structured/tabular data. The go-to model for Kaggle competitions and real-world production systems on tabular data. Needs investing time in hyperparameter tuning (learning rate, number of estimators, max depth) and interpretability is secondary to predictive power. |

**Validation Strategy:**

* **Split:** The training and test sets are already defined by the Kaggle competition. The training set will be used for model training and validation, while the test set will be used for final evaluation.
* **Cross-Validation:** A 5-fold cross-validation will be performed on the training set to evaluate the performance of each candidate model and to tune hyperparameters. This will help ensure that the model generalizes well to unseen data and prevents overfitting. Stratified K-Fold will be preferred if the Exploratory Data Analysis (EDA) reveals class imbalance in the target variable.
* **Final Evaluation:** The final model will be evaluated on the test set on the Kaggle platform, which will provide an accuracy score based on the predictions submitted.

### Scope and Constraints

* **Timeline:** No specific timeline is defined for this project, but it is expected to be completed within a reasonable timeframe (e.g., a few weeks) to allow for model development, validation, and submission.
* **Regulatory / Privacy Constraints:** The dataset used in this competition is publicly available. Even though it contains personally identifiable information (PII) such as names and ticket numbers, the data is historical and there are no specific regulatory constraints related to data privacy for this project.
* **Authorization:** This project is being conducted as part of a Kaggle competition, and all necessary permissions to use the dataset and submit predictions have been obtained through participation in the competition.
